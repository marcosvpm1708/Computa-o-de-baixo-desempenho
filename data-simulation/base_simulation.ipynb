```python
import pandas as pd
from utils import tempo_execucao
```

```python
data = pd.read_csv("../dataframe/BD_Atlas_1991_2024_Monte_Carlo.csv")
keys = data[['Protocolo_S2iD', 'Data_Registro', 'Cod_Cobrade', 'Cod_IBGE_Mun',  'regiao']]
```

```python
col_numericas = data.select_dtypes(include=['float64', 'int64']).columns.difference(['Cod_Cobrade', 'Cod_IBGE_Mun'])

def prever_crescimento_historico(df, colunas_numericas, crescimento_extra_pct):
    # Calcular média por ano
    medias_por_ano = df.groupby('Ano')[colunas_numericas].mean()

    # Calcular taxa de crescimento ano a ano
    crescimento_ano_ano = medias_por_ano.pct_change().dropna()

    # Calcular taxa média histórica de crescimento por coluna
    crescimento_medio = crescimento_ano_ano.mean()

    # Aplicar o crescimento médio + crescimento adicional fornecido
    crescimento_completo = crescimento_medio + (crescimento_extra_pct / 100)

    # Pegar último ano do dataset
    ultimo_ano = df['Ano'].max()

    # Pegar médias do último ano
    base_ano_atual = medias_por_ano.loc[ultimo_ano]

    # Projetar próximo ano com crescimento total
    previsao_proximo_ano = base_ano_atual * (1 + crescimento_completo)
    previsao_proximo_ano.name = ultimo_ano + 1

    # Empilhar tudo para retornar DataFrame final
    previsoes_df = pd.DataFrame(previsao_proximo_ano).T

    return previsoes_df, crescimento_medio

# Exemplo: prever com crescimento adicional de 5%
previsao_2026, taxa_media_historica = prever_crescimento_historico(data, col_numericas, crescimento_extra_pct=5)

# Mostrar resultados
print("📈 Crescimento médio histórico por coluna (%):")
print((taxa_media_historica * 100).round(2))

print("\n📊 Projeção para o ano de 2026 com +5% de crescimento adicional:")
print(previsao_2026.round(2))
```
